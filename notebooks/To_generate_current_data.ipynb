```python
import os
import copernicusmarine as cm
import xarray as xr

DATASET = "cmems_mod_glo_phy_my_0.083deg_P1M-m"   # courants mensuels, 1/12°
OUT_DIR = "../data"
os.makedirs(OUT_DIR, exist_ok=True)

for year in range(2017, 2022):                     # 2017 à 2021 inclus
    fichier = f"currents-{year}.nc"

    if os.path.exists(f"{OUT_DIR}/{fichier}"):     # déjà téléchargé : on saute
        print(f"{year} already exists")
        continue

    cm.subset(
        dataset_id=DATASET,
        variables=["uo", "vo"],                    # courants est et nord
        minimum_depth=0.49, maximum_depth=0.5,     # surface uniquement
        start_datetime=f"{year}-01-01",
        end_datetime=f"{year}-12-31",
        output_filename=fichier,
        output_directory=OUT_DIR,
    )

    ds = xr.open_dataset(f"{OUT_DIR}/{fichier}")
    print(year, "->", ds.sizes["time"], "mois")
    ds.close()
```
