```python
import os
import copernicusmarine as cm
import xarray as xr

DATASET = "cmems_mod_glo_phy_my_0.083deg_P1M-m"   # monthly currents, 1/12°
OUT_DIR = "../data"
os.makedirs(OUT_DIR, exist_ok=True)

for year in range(2017, 2022):                     # 2017 to 2021 inclusive
    filename = f"currents-{year}.nc"

    if os.path.exists(f"{OUT_DIR}/{filename}"):    # already downloaded: skip
        print(f"{year} already exists")
        continue

    cm.subset(
        dataset_id=DATASET,
        variables=["uo", "vo"],                    # eastward and northward currents
        minimum_depth=0.49, maximum_depth=0.5,     # surface only
        start_datetime=f"{year}-01-01",
        end_datetime=f"{year}-12-31",
        output_filename=filename,
        output_directory=OUT_DIR,
    )

    ds = xr.open_dataset(f"{OUT_DIR}/{filename}")
    print(year, "->", ds.sizes["time"], "months")
    ds.close()
```
